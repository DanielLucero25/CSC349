How did you determine which columns required missing value imputation versus other data cleaning methods, and what strategy did you choose? I used things like info(), isnull().sum(), and checks for blank strings to see where the missing values were. Most of the columns did not have missing values but TotalCharges had blank spaces that were basically acting like missing data. Since TotalCharges is numerical I used a median SimpleImputer to fill those missing values. I chose the median because it is less affected by really high or low values compared to the mean. For categorical columns, I used the most common value as the imputation strategy. I also dropped customerID because it was just an identifier and did not really help predict whether a customer would churn.

Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing and training steps. The ColumnTransformer made it easier to handle different types of columns in one place. I could scale the numerical columns and one-hot encode the categorical columns without having to do every step manually. The Pipeline then connected all of the preprocessing steps directly to the model. This made the code cleaner and made sure the same steps were being applied every time the model was trained or tested.

The pipeline also helps prevent data leakage. The preprocessing steps are learned only from the training data and then applied to the test data. This is important because if information from the test set was used during preprocessing, the model could accidentally get information it should not have seen yet. That could make the results look better than they really are.

How did comparing StandardScaler versus MinMaxScaler impact your model training, convergence, or final evaluation metrics?
I tested both StandardScaler and MinMaxScaler to see how much the scaling method affected the model. Both methods helped put the numerical features on a more similar scale, which is important because columns like tenure, MonthlyCharges, and TotalCharges have very different ranges. The final model performance was fairly similar with both scalers, so neither one caused a major change in the results. However, StandardScaler worked well with SGDClassifier because it centers the data around zero and gives the features a similar spread. This can help gradient descent train more smoothly and help the model be more consitent. MinMaxScaler also worked, but it changes the values to a fixed range between zero and one. Overall, this showed me that scaling was important for the model, even though the choice between the two methods did not make a huge difference in the final evaluation.

How does configuring SGDClassifier(loss='log_loss') differ from standard linear classification models, and what advantages do probability outputs and ROC-AUC metrics provide when evaluating customer churn?

SGDClassifier works a little differently from regular logistic regression because it trains the model using stochastic gradient descent. Instead of looking at the whole dataset at once for every update, it adjusts the model little by little while going through the training data. This can make it faster and more useful when working with larger datasets.

Using loss=log_loss makes the model work like logistic regression. One thing I liked about this setup is that it can give probability results instead of only saying whether a customer will churn or stay. For example, the model could say that one customer has a much higher chance of leaving than another customer. That would be useful for a company because they could focus more on customers who seem more likely to leave.

ROC-AUC also gives a better idea of how well the model is doing. Instead of only checking how many predictions were correct, it looks at how well the model separates customers who churn from customers who stay at different cutoff points. This is helpful because just looking at accuracy does not always tell the full story, especially when there are more customers who stay than customers who churn.
