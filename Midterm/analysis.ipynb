#Question 3:
For this part of the project, I used ColumnTransformer and Pipeline to make preprocessing my data easier and more organized. Since my dataset had both numerical values and time-related information, I couldn't treat every column the same way. I used ColumnTransformer to apply different steps depending on the type of data. For the numerical columns, I used SimpleImputer to fill in missing values and StandardScaler to scale them. For the hour column, I used OneHotEncoder to turn the hours into values that the models could understand. Using a pipeline also made it easier because I didn't have to manually repeat all these steps every time I trained a different model.

Another important reason I used a pipeline was to help prevent data leakage. I split my dataset into training and testing sets before fitting the preprocessing steps. This meant that things like the median values and scaling calculations were learned from the training data instead of the entire dataset. The same transformations were then applied to the test data without using it to calculate new values. Overall, I think using these tools made my code easier to manage and helped make my model evaluations more reliable.
